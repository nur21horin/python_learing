```r
model_slr <- lm(
  petal_length ~ sepal_length,
  data = iris_data
)

summary(model_slr)
```

```r
set.seed(123)

index <- createDataPartition(iris_data$species,
                              p = 0.8,
                              list = FALSE)

train_data <- iris_data[index, ]
test_data <- iris_data[-index, ]
```

multiple linear regression

```r
model_mlr <- lm(
  petal_length ~ sepal_length + sepal_width + petal_width,
  data = iris_data
)

summary(model_mlr)
```

```r
data <- data.frame(
  sepal_length = 5.5,
  sepal_width = 3.0,
  petal_width = 1.2
)

predicted_petal_length <- predict(model_mlr, newdata = data)

predicted_petal_length
```

Binary Logistic regression

```r
iris_binary <- iris_data %>%
  filter(species %in% c("Iris-setosa", "Iris-versicolor")) %>%
  mutate(
    is_versicolor = ifelse(species == "Iris-versicolor", 1, 0)
  )

head(iris_binary)

table(iris_binary$species)
```

Logistic regression


```r
model_logit <- glm(
  is_versicolor ~ sepal_length + sepal_width + petal_length + petal_width,
  data = iris_binary,
  family = binomial()
)

summary(model_logit)
```

```r
probs <- predict(
  model_logit,
  type = "response"
)

head(probs)
```

```r
preds <- ifelse(probs > 0.5, 1, 0)

table(
  Predicted = preds,
  Actual = iris_binary$is_versicolor
)
```

```r
confusionMatrix(
  factor(preds, levels = c(0, 1)),
  factor(iris_binary$is_versicolor, levels = c(0, 1)),
  positive = "1"
)
```

```r
new_flower <- data.frame(
  sepal_length = 5.5,
  sepal_width = 3.0,
  petal_length = 4.0,
  petal_width = 1.2
)

new_probability <- predict(
  model_logit,
  newdata = new_flower,
  type = "response"
)

new_class <- ifelse(
  new_probability > 0.5,
  "Iris-versicolor",
  "Iris-setosa"
)

new_probability
new_class
```

```r
model_eval <- lm(petal_length ~ sepal_length, data = train_data)

predictions <- predict(model_eval, newdata = test_data)

actual <- test_data$petal_length

MAE <- mean(abs(actual - predictions))
RMSE <- sqrt(mean((actual - predictions)^2))
R2 <- summary(model_eval)$r.squared

MAE
RMSE
R2
```
